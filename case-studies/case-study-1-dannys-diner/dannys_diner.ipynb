## Setup

This cell builds the three tables from `schema.sql` and defines `q()`, a shortcut for running SQL and showing the result as a table.

```python
import sqlite3
import pandas as pd

con = sqlite3.connect("dannys_diner.db")
con.executescript(open("schema.sql").read())   # builds the 3 tables

def q(sql):
    return pd.read_sql(sql, con)
```

## Warm-up: simple SELECTs

**Show all tables**

```python
sales=q("SELECT * FROM sales")
sales
```

```python
menu=q("SELECT * FROM menu")
menu
```

```python
members=q("select * FROM members")
members
```

## Question 1

What is the total amount each customer spent at the restaurant?

Join sales to menu to get each item’s price, then sum per customer.

```python
q("SELECT customer_id,sum(price) as total_spent FROM sales LEFT JOIN menu ON sales.product_id=menu.product_id \
group by customer_id ")
```
