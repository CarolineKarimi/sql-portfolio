## Question 1

**Question:** What is the total amount each customer spent at the restaurant?

**Approach:** Join sales to menu to get each item's price, then sum per customer.

```python
q("""
SELECT customer_id,sum(price) AS total_spent 
FROM sales 
LEFT JOIN menu ON sales.product_id=menu.product_id
GROUP BY customer_id
ORDER BY customer_id 
""")
```

## Question 2

**Question:** How many days has each customer visited the restaurant?

**Approach:** use groupby to count distinct days drom sales

```python
sales
```

```python
q("""
SELECT  customer_id,COUNT(DISTINCT(order_date)) AS days_visited
FROM sales GROUP BY customer_id
""")
```
